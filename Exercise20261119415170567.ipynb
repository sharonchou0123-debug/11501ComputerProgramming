```python
routes = {}

def route(path):
    def decorator(func):
        routes[path] = func
        return func
    return decorator

def dispatch(path):
    if path in routes:
        return routes[path]()
    return "404 Not Found"


@route("/")
def home():
    return "Home"

@route("/about")
def about():
    return "About"

@route("/contact")
def contact():
    return "Contact"

while True:

    path = input("URL: ")

    if path == "quit":
        break

    print(dispatch(path))
```
