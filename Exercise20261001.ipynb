2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

因為電腦記憶體有限，無法儲存無限多位的二進位數值，Python 採用 IEEE 754 標準的雙精度浮點數（64-bit Double Precision）來進行截斷與儲存：

位數限制： 64 位元中包含 1 位符號位、11 位指數位，以及 52 位尾數（Significand/Mantissa）。加上隱藏的高位 1，實際精度約為 53 個二進位位元（約 15~17 位十進位有效數字）。

捨入機制（四捨六入五成雙 / Round to nearest, ties to even）：
在截斷無限循環小數時，採用 IEEE 754 預設的捨入規則：

當末位小於半數時捨去，大於半數時進位。

當恰好在中間（Ties, 即最後剛好是 .5 情況）時，捨入到最近的偶數。這能避免傳統「四捨五入」在大量數據計算時產生的向上偏誤（Positive Bias）。

3. Python如何知道浮點數的精度？請問執行以下程式碼:
```python
import sys
print(sys.float_info)
```
並說明輸出結果的意思。

執行 import sys; print(sys.float_info) 的輸出結果類似如下（以 64 位元系統為例）

Python
sys.float_info(
    max=1.7976931348623157e+308,
    max_exp=1024,
    max_10_exp=308,
    min=2.2250738585072014e-308,
    min_exp=-1021,
    min_10_exp=-307,
    dig=15,
    mant_dig=53,
    epsilon=2.220446049250313e-16,
    radix=2,
    rounds=1,
)

輸出結果核心欄位說明：

dig=15：十進位浮點數的可靠有效位數（Precision in decimal digits）。代表在 15 位十進位數字內，轉換為浮點數再轉回不會有失真。

mant_dig=53：二進位尾數的位數（包含隱藏位），對應 IEEE 754 雙精度格式。

epsilon=2.220446049250313e-16：1 與大於 1 的最小可表示浮點數之間的差值（機器 Epsilon），代表該精度下能區分的最小相對誤差。

max & min：此系統環境能表示的最大與最小正規化正浮點數。

radix=2：浮點數表示法的基數為二進位

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
執行 print(0.1 + 0.2) 會輸出 0.30000000000000004，這是浮點數二進位近似值相加累積誤差的結果。

解決方案如下：

使用內建 decimal 模組（推薦標準解法）
decimal 模組以十進位邏輯運算，能精準表示與計算十進位小數：

Python
from decimal import Decimal

result = Decimal("0.1") + Decimal("0.2")
print(result)  # 輸出：0.3

在金融領域極為適合（最佳標準作法），但在科學計算領域通常不適合（效能與生態體系不符），在金融領域是不可或缺的最佳解法，精確性要求高、精準掌控捨入；但在科學計算領域則因效能低下與缺少底層硬體加速，並非理想的通用方法，計算效能過低、缺乏高效科學庫支援、科學關注相對誤差而非絕對精準。
